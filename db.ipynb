```python
from showrunner.models import Cue
from showrunner.database import ShowDatabase
from sqlmodel import select
```

```python
db = ShowDatabase('12AM.db')
db.create_schema()
```

```python
with db.session() as session:
    dcas = select(Cue).where(Cue.cue_type == 'DCA')  # .order_by(Cue.number)
    for dca in session.exec(dcas):
        # Change cue type to Network
        dca.cue_type = 'Network'
        session.add(dca)
    session.commit()
```
